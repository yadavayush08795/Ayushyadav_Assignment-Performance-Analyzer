## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_03_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python

# Display the shape and column names
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

# Inspect data types and missing values
print(df.info())
print("\nMissing values:\n", df.isnull().sum())

# Count duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum

# Select assignment score columns and convert them to a NumPy array
assignment_cols = ["Assignment_1", "Assignment_2", "Assignment_3", "Assignment_4"]
scores_np = df[assignment_cols].to_numpy()

# Calculate overall statistics
overall_mean = round(np.mean(scores_np), 2)
overall_max = np.max(scores_np)
overall_min = np.min(scores_np)

# Calculate average score for each assignment
col_means = np.mean(scores_np, axis=0)

# Create a summary table
summary_table = pd.DataFrame({
    "Assignment": assignment_cols,
    "Average Score": np.round(col_means, 2),
    "Maximum Score": df[assignment_cols].max().values,
    "Minimum Score": df[assignment_cols].min().values
})

# Display assignment-wise summary table
print("ASSIGNMENT PERFORMANCE SUMMARY")
display(summary_table)

# Display overall statistics in table format
overall_table = pd.DataFrame({
    "Metric": ["Overall Mean", "Overall Maximum", "Overall Minimum"],
    "Score": [overall_mean, overall_max, overall_min]
})

print("OVERALL PERFORMANCE")
display(overall_table)

```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results

# Create two new columns: Total_Marks and Average_Marks across the 4 assignments
df["Total_Marks"] = df[assignment_cols].sum(axis=1)
df["Average_Marks"] = df[assignment_cols].mean(axis=1)

# Filter: find students scoring an average of 70 or above
above_70 = df[df["Average_Marks"] >= 70]
print("Students with average >= 70:")
display(above_70[["Student_ID", "Name", "Total_Marks", "Average_Marks"]])

# Sort: rank students from highest to lowest Average_Marks
ranked_df = df.sort_values(by="Average_Marks", ascending=False)
print("\nTop 5 students:")
display(ranked_df[["Student_ID", "Name", "Total_Marks", "Average_Marks"]].head(5))

print("\nBottom 5 students:")
display(ranked_df[["Student_ID", "Name", "Total_Marks", "Average_Marks"]].tail(5))
```

## 6. Your individual project challenge
**Challenge:** Calculate total and average assignment marks; identify best performer and students with an assignment average below 40.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Identify best performer based on highest Total_Marks

# 3rd sdfghjk,l

cols = [
    "Student_ID",
    "Name",
    "Assignment_1",
    "Assignment_2",
    "Assignment_3",
    "Assignment_4",
    "Total_Marks",
    "Average_Marks",
]
best_df = df.loc[[df["Total_Marks"].idxmax()], cols]
at_risk = df.loc[df["Average_Marks"] < 40, cols]

print("Best Performer:")
print(best_df.to_string(index=False))

print(f"\nStudents with average below 40: {len(at_risk)}")
print(
    at_risk.to_string(index=False)
    if not at_risk.empty
    else "No students below 40."
)


```
